## Python End-of-Module Capstone Project: Proposal Guidelines

You are required to identify a practical problem and propose a Python application that addresses it.

Before beginning development, you must submit a project proposal for review and approval.

**Important:** All projects must use Python to develop interactive applications.

Your proposal should follow the structure below.

## Proposal Structure

### Project Title
Provide a clear and descriptive title for your application.

Example: Personal Finance and Expense Tracking Dashboard.

### Problem Statement
Explain:
- What problem are you trying to solve?
- Who experiences this problem?
- Why is it important to address?
- How will your application help solve it?

**Example:**

Many university students and young professionals struggle to track their daily expenses and manage monthly budgets. This makes it difficult to understand spending patterns and control unnecessary expenditure.

This project proposes a Personal Finance Dashboard that allows users to record income and expenses, monitor budgets, and visualize their spending habits.

### Project Objectives
Provide one general objective and three to five specific objectives.

**Example:**

General Objective: To develop an interactive application for managing personal finances.

Specific Objectives:
- Allow users to record and categorize income and expenses.
- Enable users to set monthly budgets.
- Calculate total income, expenditure, and remaining balances.
- Visualize spending patterns using interactive charts.

### Proposed Solution
Describe the application you intend to build, its intended users, and how it will address the identified problem.

**Example:**

A web-based application developed using Python and Streamlit. Users will record transactions, set budgets, and view financial summaries through an interactive dashboard.

### Proposed Application Functionalities
List and briefly explain at least four core features your application will implement.

**Example:**

| Functionality | Description |
|---|---|
| Record Transactions | Capture income and expenses. |
| Categorize Expenses | Group expenses into categories. |
| Budget Management | Set and monitor monthly budgets. |
| Financial Dashboard | Display income, expenses, and balances. |
| Spending Analysis | Visualize expenditure using charts. |
| Export Reports | Download transaction records. |

### Data Requirements and Sources
Explain what data your application requires and where it will come from.

Possible sources include:
- Web scraping.
- APIs.
- User input through application forms.
- Manually collected records.

For external datasets, provide the dataset name, source link, format, and how you intend to use it.

If your application relies on user input, explain what information users will provide and how it will be stored.

### Proposed Technology Stack
Identify the technologies and libraries you intend to use and briefly explain their purpose.

| Technology | Purpose |
|---|---|
| Python | Application development. |
| Streamlit | Building the web interface. |
| Pandas | Data manipulation and analysis. |
| Plotly / Matplotlib | Data visualization. |
| CSV / JSON / SQLite | Data storage. |
| Git and GitHub | Version control and project hosting. |

### Project Scope
Explain what you intend to implement and what you will leave out.

For example:

- **Within scope:** Recording transactions, setting budgets, and generating financial summaries.
- **Outside scope:** Direct bank integration, automatic M-Pesa synchronization, and mobile banking integration.

Keep your project achievable within the available timeline.

## Python Concepts to Demonstrate

Your application must provide opportunities to apply the Python concepts covered in this module, including:

- Variables, data types and control flow.
- Lists, tuples, dictionaries and sets.
- Functions.
- Object-Oriented Programming (at least one user-defined class).
- File handling.
- Exception handling and input validation.
- Working with databases.

## Submission Requirements

Submit a proposal document containing all the sections outlined above.

Before submitting, ensure that:
- Your problem is clearly defined.
- Your objectives align with your proposed solution.
- Your application functionalities are specific and achievable.
- Your data sources are identified, where applicable.
- Your technology stack includes Python and Streamlit.
- Your project scope is realistic.

**Important:** Submit your proposal for review and approval before starting full development. Your proposal should explain what you intend to build, not present a completed application.
