## What Are Regular Expressions?

Imagine you have a dataset containing thousands of customer records. Each record contains a name, email address, phone number, and other information.

You are asked to extract all the phone numbers.

You could search for a specific phone number using ordinary Python string methods. However, you do not know the actual phone numbers in advance.

What you need is a way to describe what a phone number looks like.

This is where **Regular Expressions (Regex)** become useful.

`Regular Expressions` are patterns used to search, match, extract, validate, and manipulate text.

Instead of searching for one exact value, Regex allows us to describe a pattern that different text values can follow.

## Why Do We Need Regex?

Python provides string methods such as `find()`, `replace()`, `split()`, and the `in` operator.

These are useful when we know exactly what we are looking for.

For example:

```python
text = "I am learning Python"
print("Python" in text)
```

But suppose we have the following text:

```python
text = "Mary scored 85, John scored 90 and Ann scored 78"
```

We want to extract all the scores without searching for `85`, `90`, or `78` individually.

Instead, we can describe what a score looks like: a sequence of digits.

```python
import re
```

```python
scores = re.findall(r"\d+", text)
print(scores)
```

## Where Is Regex Used?

Regex is useful in many areas of programming and data analysis.

| Application | Example |
|---|---|
| Data extraction | Extracting emails, phone numbers, and dates |
| Data cleaning | Removing unwanted symbols and characters |
| Data validation | Checking whether input follows a required format |
| Text analysis | Finding specific words or patterns |

Regex is not limited to Python. It is also used in JavaScript, Java, SQL databases, text editors, and command-line tools.

### Understanding Raw Strings 
Before writing Regex patterns, we need to understand something important about backslashes (`\`).

Python uses backslashes to represent special characters:

- `\n` represents a new line.
- `\t` represents a tab.

Regex also uses backslashes to represent special patterns:

- `\d` represents a digit.
- `\w` represents a word character.
- `\s` represents whitespace.

This can create confusion because Python may interpret a backslash before the Regex engine processes the pattern.

### What Is a Raw String?

A raw string is a string in which Python preserves backslashes instead of interpreting them as escape sequences.

We create one by placing `r` before the quotation marks.

Compare:

```python
print("\n")
print(r"\n")
```

The first string represents a new line, while the second contains the literal characters `\` and `n`.

Now consider a Regex pattern:

```python
pattern = r"\d"
```

The `r` tells Python to preserve the backslash so that the Regex engine can interpret `\d` as a digit.

**Important:** Raw strings do not remove the special meaning of Regex characters. They simply make it easier to write patterns containing backslashes.

For this lesson, we will use raw strings whenever we write Regex patterns.

# Understanding Regex Patterns

A Regex pattern is made up of characters and symbols that describe what we want to find.

There are two important things to understand:

1. Literal characters match themselves.
2. Special Regex symbols have specific meanings.

## Literal Characters

Literal characters are ordinary characters that match exactly what they represent.

For example:

```python
text = "I am learning Python"

result = re.findall(r"Python", text)

print(result)
```

The pattern `Python` looks for that exact sequence of characters.

Let's try another example:

```python
text = "Python is popular. I use Python every day."

result = re.findall(r"Python", text)

print(result)
```

Notice that the pattern finds both occurrences.

## Metacharacters

So far, we have searched for exact words.

But what if we want to find text that follows a particular structure rather than an exact word?

Regex provides special symbols called **metacharacters**.

Metacharacters allow us to describe more flexible patterns.

`Metacharacters` are special characters in Regular Expressions (Regex) that have a special meaning. They help us create patterns to search for specific characters, words or text.

We will introduce them gradually.

### **Common Regex Metacharacters**

| Metacharacter | Meaning                                                       | Example     | What it matches           |
| ------------- | ------------------------------------------------------------- | ----------- | ------------------------- |
| `.`           | Matches any single character except a newline.                | `c.t`       | cat, cut, cot             |
| `[]`          | Matches any one character inside the brackets.                | `[abc]`     | a, b, or c                |
| `[^...]`      | Matches any character not listed inside the brackets.         | `[^abc]`    | d, e, f, etc.             |
| `()`          | Groups parts of a pattern together.                           | `(cat)`     | cat                       |
| `\\|`         | Matches either pattern (OR).                                  | `cat\\|dog` | cat or dog                |
| `\`           | Escapes a special character or introduces a special sequence. | `\.`        | A literal full stop (`.`) |

Note: To match a metacharacter as an ordinary character, use a backslash. For example, `\.` matches an actual full stop.

### The Dot `(.)`

The dot matches any single charachter except a newline

```python
text = "cat, cot, caat, cut, bat, mat"

result = re.findall(r"c.t", text)

print(result)
```

### Escaping Special Characters

Sometimes we want to search for a character that Regex normally treats as special.

For example, the dollar sign `$` has a special meaning in Regex. It represents the end of a string.

If we want to find an actual dollar sign, we must escape it using a backslash.

```python
text = "The products cost $20, $35 and $50."

result = re.findall(r"\$", text)

print(result)
```

Similarly, the dot `.` normally matches any single character. To find an actual full stop, we escape it:

```python
text = "The price is 250.50."

result = re.findall(r"\.", text)

print(result)
```

The backslash tells Regex to treat the special character literally.

## Character Classes

So far, we have used the dot to match any single character.

But sometimes we want to be more specific.

For example, suppose we want to match a word that starts with `c` and ends with `t`, but the middle character must be either `a` or `o`.

We can use **character classes**.

Character classes allow us to specify exactly which characters should be matched.

They are written using square brackets `[]`.

```python
text = "cat, cot, caat, cut, bat, mat"

result = re.findall(r"c[ao]t", text)
print(result)
```

Let's break down the pattern:

- `c` means the text must begin with `c`.
- `[ao]` means the next character can be either `a` or `o`.
- `t` means the text must end with `t`.

Square brackets match one character from the options provided.

Unlike the dot, which matches any single character, `[ao]` only matches `a` or `o`.

## Character Ranges

Instead of listing every character individually, we can specify a range.

| Pattern | Meaning |
|---|---|
| `[a-z]` | Any lowercase letter |
| `[A-Z]` | Any uppercase letter |
| `[0-9]` | Any digit |
| `[a-zA-Z]` | Any uppercase or lowercase letter |
| `[a-zA-Z0-9]` | Any letter or digit |

```python
text = "Mary123"

letters = re.findall(r"[a-zA-Z]", text)
digits = re.findall(r"[0-9]", text)

print(letters)
print(digits)
```

The pattern `[a-zA-Z]` matches one letter at a time, whether uppercase or lowercase.

The pattern `[0-9]` matches one digit at a time.

## Negated Character Classes (`[^...]`)

Sometimes we want to match everything except certain characters.

We use the caret `^` inside square brackets to exclude characters.

```python
text = "abc123"

result = re.findall(r"[^0-9]", text)

print(result)
```

The pattern `[^0-9]` means match any character that is not a digit.

Notice the difference:

- `[0-9]` matches digits.
- `[^0-9]` matches characters that are not digits.

The meaning of `^` depends on where it appears:

- Inside square brackets at the beginning, it means NOT.
- Outside square brackets, it represents the beginning of a string.

## Common Shortcuts for Character Classes

Regex provides shortcuts for commonly used character classes.

These shortcuts make patterns shorter and easier to write.

| Pattern | Meaning |
|---|---|
| `\d` | Any digit |
| `\D` | Any non-digit |
| `\w` | Any word character: letters, digits, or underscore |
| `\W` | Any non-word character |
| `\s` | Any whitespace character, such as a space, tab, or newline |
| `\S` | Any non-whitespace character |

Let's see how they work:

```python
text = "Mary scored 85 in Python."

# Find digits
print(re.findall(r"\d", text))

# Find groups of word characters
print(re.findall(r"\w+", text))

# Find whitespace characters
print(re.findall(r"\s", text))
```

Notice the difference between `\d` and `\w+`:

- `\d` matches each digit individually.
- `\w+` matches groups of consecutive word characters.

## Quantifiers

Character classes help us define which characters we want to match.

But what if we want to specify how many times a character or pattern should appear?

For example:

- A number may contain several digits.
- A username may contain between 5 and 12 characters.
- A word may contain one or more letters.

Regex uses **quantifiers** to specify how many times a character or pattern should occur.

### Common Quantifiers

| Quantifier | Meaning | Example |
|---|---|---|
| `*` | Zero or more occurrences | `ab*` |
| `+` | One or more occurrences | `ab+` |
| `?` | Zero or one occurrence | `colou?r` |
| `{n}` | Exactly n occurrences | `\d{4}` |
| `{n,}` | At least n occurrences | `\d{2,}` |
| `{n,m}` | Between n and m occurrences | `\d{2,4}` |

## Zero or More (`*`)

The `*` quantifier matches zero or more occurrences of the preceding character or pattern.

```python
text = "ac, abc, abbc, abbbc"

result = re.findall(r"ab*c", text)

print(result)
```

The pattern `ab*c` means:

- `a` must appear.
- `b*` means zero or more `b` characters.
- `c` must appear.

This is why even `ac` matches: it contains zero `b` characters.

## One or More (`+`)

The `+` quantifier matches one or more occurrences.

```python
text = "ac, abc, abbc, abbbc"

result = re.findall(r"ab+c", text)

print(result)
```


Unlike `*`, the `+` quantifier requires at least one `b`.

## Zero or One (`?`)

The `?` quantifier makes the preceding character optional.

For example, the words `color` and `colour` differ by one optional character.

```python
text = "color and colour"

result = re.findall(r"colou?r", text)

print(result)
```

The `u?` means that `u` may appear zero or one time.

## Exact and Range-Based Repetition

We can use curly brackets `{}` when we want to specify the number of repetitions.

```python
text = "7 45 123 2026 12345"

# Exactly four digits
result = re.findall(r"\d{4}", text)

print(result)
```

```python
text = "7 45 123 2026 12345"

print(re.findall(r"\d{2,}", text))
print(re.findall(r"\d{2,4}", text))
```

The first pattern matches sequences containing at least two digits.

The second pattern matches between two and four digits. It does not require the entire number to contain no additional digits; it finds matching portions within the text.

## Anchors and Word Boundaries

Sometimes we do not just want to find a pattern anywhere in a piece of text.

We may want to specify where the pattern should appear.

Anchors specify the position where a pattern should match. They do not match actual characters.

For example:

- A username must start with a letter.
- A line must end with a full stop.
- A word must appear as a complete word rather than part of another word.

Regex uses **anchors** and **word boundaries** to describe these positions.

### **Common Regex Anchors**

| Anchor | Meaning                            | Example  | What it matches |
| ------ | ---------------------------------- | -------- | --------------- |
| `^`    | Matches the beginning of a string. | `^Hello` | Hello world     |
| `$`    | Matches the end of a string.       | `world$` | Hello world     |

### Beginning of a String (`^`)

The `^` symbol matches the beginning of a string.

```python
text = "Python is easy to learn."

result = re.findall(r"^Python", text)

print(result)
```

The pattern only matches `Python` if it appears at the beginning.

## End of a String (`$`)

The `$` symbol matches the end of a string.

```python
text = "I am learning Python"

result = re.findall(r"Python$", text)

print(result)
```

## Word Boundary (`\b`)

The `\b` symbol represents a boundary between a word character and a non-word character, or the beginning or end of a string.

It is useful when we want to find a complete word rather than a sequence of characters inside another word.

```python
text = "I use Python. Python is powerful. I also like Python programming."

result = re.findall(r"\bPython\b", text)

print(result)
```

The pattern `\bPython\b` looks for `Python` as a complete word.

## Difference Between `Python` and `\bPython\b`

```python
text = "Python and Pythonic are related."

print(re.findall(r"Python", text))
print(re.findall(r"\bPython\b", text))
```

## Common Regex Functions in Python

Now that we understand how to construct Regex patterns, let's explore the functions Python provides for working with them.

The `re` module provides several functions. Each serves a different purpose.

## Common Regex Functions

| Function | Purpose | Returns |
|---|---|---|
| `re.findall()` | Finds all matches | A list |
| `re.search()` | Finds the first match anywhere in the text | A Match Object or `None` |
| `re.match()` | Checks for a match at the beginning of the text | A Match Object or `None` |
| `re.fullmatch()` | Checks whether the entire text matches the pattern | A Match Object or `None` |
| `re.finditer()` | Finds all matches and returns them one at a time | An iterator of Match Objects |
| `re.sub()` | Replaces matching text | A new string |
| `re.split()` | Splits text wherever the pattern matches | A list |
| `re.compile()` | Creates a reusable Regex pattern | A compiled pattern object |
