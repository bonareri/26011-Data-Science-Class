# Introduction to Data Analysis, Data Science and Python

## Instructor Teaching Guide

### Lesson Overview

This lesson introduces students to Data, Data Analysis, Data Science and the role of Python in these fields. Students should understand why organisations collect data, how data is used to solve real-world problems and why Python is an important tool for working with data.

The lesson should establish the context for learning Python so that students understand how the programming concepts they will learn connect to practical applications in Data Analysis and Data Science.

### Learning Objectives

By the end of this lesson, students should be able to:

1. Define data and explain its importance.
2. Explain what Data Analysis is and why organisations analyse data.
3. Describe the Data Analysis process.
4. Define Data Science and identify its applications.
5. Differentiate between Data Analysis and Data Science.
6. Explain the importance of Python in Data Analysis and Data Science.
7. Identify common Python libraries used in data-related tasks.
8. Explain how Python can be used to solve real-world problems.

# PART 1: UNDERSTANDING DATA

## 1. What is Data?

**Content to teach:**

Data refers to raw facts, observations, measurements or information collected about people, objects, events or activities.

Examples include:

- The age of students in a classroom.
- Daily sales recorded by a supermarket.
- Monthly electricity consumption.
- Customer purchases on an e-commerce platform.
- Patient records in a hospital.
- Rainfall measurements collected over several years.

Data is generated everywhere. Every time someone makes a purchase, searches online, uses mobile money, visits a hospital or interacts with a digital platform, data is generated.

However, raw data alone is not necessarily useful. We need ways of processing, analysing and interpreting it to extract meaningful information.

**Instructor guidelines:**

- Begin by asking students to identify examples of data they interact with daily.
- Use familiar examples such as M-Pesa transactions, social media interactions, online purchases and school attendance records.
- Explain that data can be numerical, categorical, textual, visual or time-based.
- Emphasise that data becomes valuable when we can extract useful information from it.

**Questions to ask students:**

1. What information does M-Pesa collect when someone sends money?
2. What data might a supermarket collect from its customers?
3. Why do organisations collect data?
4. Is raw data automatically useful? Why or why not?

# PART 2: INTRODUCTION TO DATA ANALYSIS

## 2. What is Data Analysis?

**Content to teach:**

Data Analysis is the process of collecting, cleaning, transforming, exploring and interpreting data to discover meaningful patterns, answer questions, draw conclusions and support decision-making.

In simpler terms:

**Data Analysis is about using data to understand what is happening, why it is happening and what decisions we can make based on the evidence.**

### Practical Example: Supermarket Sales

Imagine you are working for a supermarket.

The supermarket has collected sales data containing:

- Products sold.
- Quantity sold.
- Prices.
- Dates of transactions.
- Branch locations.

The management wants to understand its business performance.

As a Data Analyst, you might investigate:

1. Which products generate the highest revenue?
2. Which branch makes the most sales?
3. Which months experience the highest sales?
4. Why are sales declining in a particular branch?
5. What can the business do to improve sales?

You would use data to answer these questions and communicate your findings to management.

**Instructor guidelines:**

- Walk students through the supermarket example.
- Ask them to imagine they have been hired as Data Analysts.
- Explain that their responsibility is not simply to look at numbers but to investigate business questions.
- Emphasise the importance of understanding the problem before starting the analysis.

**Questions to ask students:**

1. If you were the supermarket manager, what would you want to know about your business?
2. What data would you need to answer those questions?
3. How would the findings help the supermarket make decisions?

## 3. The Data Analysis Process

Data Analysis generally involves the following steps:

### 3.1 Data Collection

Gathering relevant data from different sources.

Examples:
- Surveys.
- Databases.
- Websites.
- APIs.
- Spreadsheets.
- Transaction systems.

### 3.2 Data Cleaning

Identifying and correcting issues in the dataset.

Examples:
- Handling missing values.
- Removing duplicates.
- Correcting inconsistent entries.
- Identifying incorrect data types.

### 3.3 Data Transformation

Preparing data into a suitable format for analysis.

Examples:
- Converting dates into appropriate formats.
- Creating new variables.
- Combining datasets.
- Standardising values.

### 3.4 Data Exploration

Investigating patterns, trends and relationships within the data.

Examples:
- Identifying the most frequently purchased products.
- Examining sales trends.
- Comparing performance across branches.
- Identifying relationships between variables.

### 3.5 Data Analysis

Applying statistical methods to answer questions and investigate relationships.

Examples:
- Calculating averages.
- Measuring variability.
- Correlation analysis.
- Hypothesis testing.

### 3.6 Data Visualisation and Communication

Presenting findings through charts, dashboards and reports.

Examples:
- Bar charts.
- Line graphs.
- Histograms.
- Interactive dashboards.

**Instructor guidelines:**

Explain that these steps are interconnected and that data analysis is not always a strictly linear process. Analysts may return to earlier stages when they discover data quality issues or new questions.

Emphasise that the ultimate goal is to generate insights that support decision-making.

**Questions to ask students:**

1. Why do we need to clean data before analysing it?
2. What could happen if we analyse incorrect or duplicated records?
3. Why are visualisations useful when presenting findings?

# PART 3: INTRODUCTION TO DATA SCIENCE

## 4. What is Data Science?

**Content to teach:**

Data Science is an interdisciplinary field that combines programming, mathematics, statistics, domain knowledge and computational techniques to extract insights from data, build predictive models and solve complex problems.

While Data Analysis often focuses on understanding existing data and answering questions, Data Science goes further by using data to develop models and systems that can learn patterns, make predictions and support automated decisions.

### Practical Example: Fraud Detection

Imagine a bank wants to identify fraudulent transactions.

It has millions of historical transactions, some legitimate and others fraudulent.

A Data Scientist can:

1. Collect historical transaction data.
2. Clean and prepare the data.
3. Explore patterns associated with fraudulent transactions.
4. Apply machine learning algorithms.
5. Train a model to recognise suspicious transaction patterns.
6. Evaluate the model's performance.
7. Deploy the model to help identify potentially fraudulent transactions in real time.

### Other Applications of Data Science

- Predicting diseases using medical data.
- Detecting spam and fraudulent messages.
- Recommending products on e-commerce platforms.
- Predicting customer behaviour.
- Forecasting electricity demand.
- Building chatbots and intelligent systems.

**Instructor guidelines:**

- Explain that Data Science combines multiple disciplines.
- Introduce the concepts of statistics, programming, mathematics and domain knowledge.
- Explain machine learning at a high level without going into technical implementation.
- Use familiar applications such as Netflix recommendations, fraud detection and weather forecasting.
- Help students understand that Data Science is not just about building machine learning models; it involves the entire process of working with data.

**Questions to ask students:**

1. How does Netflix recommend movies to its users?
2. How can a bank identify suspicious transactions?
3. What is the difference between analysing previous transactions and predicting future fraudulent transactions?

# PART 4: DATA ANALYSIS VS DATA SCIENCE

## 5. Difference Between Data Analysis and Data Science

| Data Analysis | Data Science |
|---|---|
| Focuses on understanding data and generating insights. | Focuses on extracting insights and building predictive or intelligent systems. |
| Often answers what happened and why. | Often investigates what might happen and how systems can learn from data. |
| Uses descriptive and inferential statistics. | Uses statistics, machine learning and computational methods. |
| Uses Excel, SQL, Python, Power BI and Tableau. | Uses Python, SQL, machine learning frameworks and other computational tools. |
| Produces reports, dashboards and recommendations. | Produces predictive models, intelligent systems and data-driven applications. |

**Important:** Data Analysis and Data Science overlap. Data Science builds on many of the foundations students learn in Data Analysis.

**Instructor guidelines:**

Use the supermarket and fraud detection examples to illustrate the difference.

Emphasise that both fields require the ability to understand data, clean it, explore it and interpret results.

Do not present the fields as completely separate. Explain that Data Science incorporates many Data Analysis techniques.

# PART 5: INTRODUCTION TO PYTHON

## 6. What is Python?

Python is a high-level, general-purpose programming language known for its readable syntax, versatility and extensive ecosystem of libraries.

It is widely used in:

- Data Analysis.
- Data Science.
- Machine Learning.
- Artificial Intelligence.
- Automation.
- Software Development.

## 7. Why is Python Important in Data Analysis and Data Science?

**Instructor introduction:**

Now that we understand Data Analysis and Data Science, we need to understand the tools that enable us to work with data.

One of the most widely used tools in these fields is Python.

You might wonder why we need Python when we already have tools such as Excel.

Excel is useful for working with structured data, performing calculations and creating reports. However, when datasets become larger, analytical tasks become more complex or processes need to be automated, programming languages such as Python become extremely useful.

### 7.1 Data Manipulation and Cleaning

Python allows us to:

- Clean datasets.
- Handle missing values.
- Remove duplicates.
- Filter records.
- Transform data.
- Combine multiple datasets.

Libraries such as Pandas and NumPy make these operations easier.

### 7.2 Data Visualisation

Python allows us to create charts, graphs and visualisations to understand patterns and communicate findings.

Libraries include:

- Matplotlib.
- Seaborn.

### 7.3 Statistical Analysis

Python allows us to perform:

- Statistical calculations.
- Hypothesis testing.
- Correlation analysis.
- Descriptive statistics.
- Inferential statistics.

### 7.4 Machine Learning

Python allows us to build models that learn from historical data and make predictions.

Examples include:

- Predicting house prices.
- Detecting fraudulent transactions.
- Predicting customer churn.
- Classifying emails as spam or legitimate.

A popular library for this is Scikit-learn.

### 7.5 Automation

Python helps automate repetitive tasks such as:

- Processing files.
- Generating reports.
- Extracting data.
- Performing repetitive calculations.
- Updating datasets.

### 7.6 Artificial Intelligence

Python provides tools and frameworks for developing:

- Deep learning models.
- Natural Language Processing applications.
- Neural networks.
- Intelligent systems.
- AI-powered applications.

**Instructor guidelines:**

- Explain that Python is a tool, not Data Analysis or Data Science itself.
- Compare Excel and Python using a simple example of analysing sales data.
- Explain that Python allows analysts to automate repetitive tasks and create reusable workflows.
- Avoid overwhelming students with library documentation at this stage.
- Emphasise that students will gradually learn how these capabilities work.

# PART 6: PYTHON LIBRARIES

## 8. Python Libraries Used in Data Analysis and Data Science

Python has a large ecosystem of libraries that extend its capabilities.

| Library | Purpose |
|---|---|
| NumPy | Numerical computations and working with arrays. |
| Pandas | Data manipulation and analysis. |
| Matplotlib | Creating data visualisations. |
| Seaborn | Statistical data visualisation. |
| Scikit-learn | Machine learning. |
| TensorFlow | Deep learning and neural networks. |
| PyTorch | Deep learning and neural networks. |

**Important distinction:**

Python is the programming language, while libraries provide additional functionality that makes specialised tasks easier.

**Instructor guidelines:**

Introduce the libraries and their purposes without teaching their syntax yet.

Explain that students will encounter these libraries as they progress through the course.

For example, when they learn Pandas, they will understand how Python can be used to work with actual datasets.

# PART 7: BRINGING EVERYTHING TOGETHER

## 9. Real-World Application: Malaria Prediction in Kenya

Imagine the Ministry of Health wants to understand malaria cases across different counties in Kenya and predict areas likely to experience increased cases.

A Data Analyst, Data Scientist and Python developer could approach this problem in the following ways:

| Area | What happens |
|---|---|
| Data Analysis | Analyse historical malaria cases to identify counties with the highest cases and seasonal trends. |
| Data Science | Develop predictive models to estimate future malaria cases. |
| Python | Used to clean the data, perform statistical analysis, visualise patterns and develop machine learning models. |

This demonstrates that Python is a tool that enables us to carry out both Data Analysis and Data Science.

**Instructor guidelines:**

Walk students through the scenario and explain how the different concepts connect.

Emphasise that real-world problems often require more than one tool or discipline.

The objective is to help students understand how the programming concepts they will learn can eventually be applied to meaningful problems.

# PART 8: CLASSROOM DISCUSSION AND ACTIVITIES

## 10. Discussion Questions

Before starting Python syntax, ask students:

1. Where do you think companies get their data?
2. Why do you think organisations collect so much data?
3. Can raw data help an organisation make decisions without being analysed?
4. What is the difference between analysing historical sales and predicting future sales?
5. Why might Python be useful when working with thousands or millions of records?

Encourage students to share their thoughts and provide examples.

## 11. Practical Discussion Activity

**Scenario:**

A mobile money company wants to understand customer transactions and identify suspicious activities.

Ask students to discuss the following:

1. What data would the company need?
2. What questions could a Data Analyst investigate?
3. What could a Data Scientist build?
4. How could Python help solve this problem?

### Expected Discussion Points

Students may identify:

- Transaction amounts.
- Transaction dates and times.
- Customer transaction histories.
- Transaction locations.
- Frequency of transactions.
- Failed transactions.
- Suspicious transaction patterns.

A Data Analyst could investigate transaction patterns and identify unusual behaviour.

A Data Scientist could develop a machine learning model to classify potentially fraudulent transactions.

Python could be used to clean the data, explore patterns, train models and automate parts of the detection process.

**Instructor guidelines:**

- Give students 5–10 minutes to discuss the scenario.
- Encourage different students to share their answers.
- Do not expect them to know technical implementation details.
- Focus on their ability to identify the problem, relevant data and possible solutions.

# PART 9: TRANSITION INTO PYTHON PROGRAMMING

## 12. Why Are We Starting With Python Fundamentals?

Now that we understand what Data Analysis and Data Science are, and why organisations rely on data to make decisions, we need to understand how we can work with that data.

One of the tools we will use throughout this course is Python.

We will start by learning the fundamentals of Python:

1. Variables.
2. Data types.
3. Operators.
4. Conditional statements.
5. Loops.
6. Functions.

You might initially wonder why we need to learn these programming concepts when our ultimate goal is to analyse data.

The reason is that before we can manipulate datasets, perform statistical analysis or build machine learning models, we need to understand how to write instructions that a computer can execute.

Think of Python fundamentals as the foundation. Everything we learn later, including Pandas, NumPy, data visualisation and machine learning, builds on these concepts.

**By the end of this journey, the goal is not simply to write Python code. It is to use Python to solve real-world problems using data.**

### Instructor Closing Guidelines

Before beginning the first Python lesson:

- Explain where students will use Python throughout the course.
- Reassure beginners that they do not need prior programming experience.
- Emphasise the importance of practising consistently.
- Encourage students to understand concepts rather than memorise code.
- Explain that making errors is part of learning programming.
- Encourage students to experiment, debug and ask questions.

1. **Data Analysis:** Helps us understand data, identify patterns and make informed decisions.
