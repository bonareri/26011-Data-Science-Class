## Strings

A `string` is a sequence of characters. It can contain letters, numbers, spaces, and symbols.

We use strings whenever a program needs to work with text, such as:

- Student names
- Email addresses
- Phone numbers
- Product descriptions
- Customer feedback
- Dates stored as text
- Messages and search terms

A string is created by placing text inside quotation marks.

Both single and double quotation marks can be used. The important thing is to open and close the string using matching quotation marks.

````python
student_name = "Mary"
course = "Data Analytics"
phone_number = "+254717676543"
print(student_name)
print(course)
print(phone_number)
````

### Why are phone numbers often stored as strings?

A phone number may contain digits, but we do not normally perform arithmetic on it. We also need to preserve leading zeros and symbols such as `+`.

```python
phone_number = "0712345678"

print(phone_number)
```

If we stored the phone number as an integer, a leading zero could be lost.

### String Indexing

Indexing allows us to access one character at a particular position in a string.

Python uses zero-based indexing. This means the first character is at index `0`, not index `1`.

````python
student_name = "Mary"

print(student_name[0])

print(student_name[1])

print(student_name[2])

print(student_name[3])
````

### When is indexing useful?

Imagine that a system needs to inspect the first character of a student ID or access the last character of a code.

````python
student_id = "STU204"

first_character = student_id[0]

last_character = student_id[-1]

print(first_character)

print(last_character)
````

## String Slicing

Indexing accesses one character. Slicing extracts a section of a string.

The general syntax is:

```python
string[start:stop]
```

- `start` is the index where extraction begins.
- `stop` is the index where extraction ends, but the character at this index is not included.

````python
course = "Python"

print(course[0:2])

print(course[0:4])

print(course[2:6])
````

### Leaving out the start or stop

If the start is omitted, Python starts from the beginning.

If the stop is omitted, Python continues to the end.

````python
course = "Python"

print(course[:3])

print(course[2:])

print(course[:])
````

### Using a step

Slicing can also include a step:

```python
string[start:stop:step]
```

The step determines how many positions Python moves forward between selected characters.

````python
word = "Programming"

print(word[0:10:2])

print(word[::2])
````

A negative step can be used to move backwards.

````python
word = "Python"

print(word[::-1])
````

## Strings Are Immutable

Strings are immutable. This means that once a string has been created, its individual characters cannot be changed directly.

````python
student_name = "Mary"

student_name[0] = "H"
````

This raises a `TypeError` because Python does not allow direct modification of a character in a string.

### How do we change a string, then?

````python
student_name = "Mary"

student_name = "H" + student_name[1:]

print(student_name)
````

The original string was not edited character by character. A new string was created and assigned to `student_name`.

### Why does immutability matter?

- It prevents accidental changes to existing text.
- It helps explain why string methods often need to be assigned to a variable.
- It encourages us to distinguish between the original text and the cleaned version.


### String Travesal

`Traversal` means visiting each character in a string one at a time

We can use a `for` loop

````python
course = "Python"

for character in course:

    print(character)
````

````python
course = "python"
print(len(course))
````

### Traversing using indexes and `range()`

Sometimes we need the position of each character as well as the character itself. We can use `range()` and `len()`.

````python
course = "Python"

for index in range(len(course)):

    print(index, course[index])
````

### Counting a character

````python
response  = "Data analysis is about finding patterns "

count =  0

for character in response:
    if character == "a":
        count += 1

print(count)
````

### String Comparison

Strings can be compared using comparison operators such as:

- `==` equal to
- `!=` not equal to
- `<` less than
- `>` greater than

String equality is case-sensitive. This means `"Python"` and `"python"` are considered different strings.

````python
first_word = "Python"
second_word = "python"

print(first_word == second_word)
print(first_word != second_word)
````

## Membership Operators

Membership operators check whether a character or sequence of characters exists inside a string.

- `in` checks whether something is present.
- `not in` checks whether something is absent.

````python
email = "mary@example.com"

print("@" in email)

print("gmail" in email)

print("yahoo" not in email)
````

## Replacing Text

`replace()` replaces one part of a string with another.

Syntax:

```python
string.replace(old, new)
```

````python
message = "I am learning Java"

updated_message = message.replace("Java", "Python")

print(updated_message)
````

We can also replace unwanted characters.

````python
phone_number = "0712-345-678"

clean_number = phone_number.replace("-", "")

print(clean_number)
````

Remember: `replace()` returns a new string.

## File Handling

### Why Do we Need File Handling?

File handling allows a python program to:
- Read information that has already been saved
- Create new files
- Write information into the files
- Add new information without deleting existing content

### Opening a File

Python uses the `open()` function to open a file so that we can read and write information

**Syntax:**
 
```python
file = open("filename.txt", "mode")
```

The `mode` determines how we interact with the file

### Common File Modes

| Mode | Meaning |
|---|---|
| `"r"` | Read an existing file |
| `"w"` | Write to a file; creates it if it does not exist and overwrites existing content |
| `"a"` | Append information to the end of a file without deleting existing content |
| `"x"` | Create a new file; raises an error if it already exists |

`NB:` Be careful when using `w` because it replaces the existing content

### Wring to a File

We use `w` mode and the `write()` function to write information into a file

````python
student_name = input("Enter student name: ").strip().title()

with open("students.txt", "w") as file:
    file.write(student_name + "\n")

print("Student registered successfully")   
````

### Reading from a File

We use `r` mode and the `read()` function to retrieve information from an existing file


````python
with open("students.txt", "r") as file:
    content = file.read()
    print(content)
````

### Appending to a File

We use the `a` mode to add content to a file without deleting existing records

````python
student_name = input("Enter student name: ").strip().title()

with open("students.txt", "a") as file:
    file.write(student_name + "\n")

print("Student registered successfully")  
````

### Different Ways to Read a File

| Method | Purpose | Returns |
|---|---|---|
| `read()` | Reads the entire file | String |
| `readline()` | Reads one line | String |
| `readlines()` | Reads all lines | List of strings |

````python
# Read the first line
with open("students.txt", "r") as file:
    print(file.readline())
````

````python
# Read all lines
with open("students.txt", "r") as file:
    print(file.readlines())
````

### Reading a File Line by Line

We can use a loop to read the records individually

````python
with open("students.txt", "r") as file:
    for students in file:
        print(students.strip())
````

### Closing a File

````python
file = open("students.txt", "r")
content = file.read()
print(content)
file.close()
````
