## Lists

A `list` is an ordered and changeable collection of elements

Lists allow duplicate values and can store different data types

### Creating a List

Lists are created using square brackets `[]`

````python
students = ["Mary", "Brian", "John", "Faith", "Peter"]

scores = [85, 90, 75, 85]

mixed = ["Mary", 22, True, 65.6]
````

### Characteristics of List

- Ordered - Elements maintain their order of insertion
- Mutable - Elements can be added, remove or changed
- Allow duplicates
- Indexed - Elements can be accessed using their positions
- Allow different data types

````python
students = ["Mary", "Brian", "John", "Faith", "Peter"]

students[-2]
````

- Positive indexing starts from `0`
- Negative indexing starts at `-1` from end


### List Slicing

Slicing allows us to access a portion of a list 

**Syntax**
```python
list_name[start:stop:step]
```

The `stop` position is excluded

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

# numbers[1:5]
#numbers[1:5:2]
# numbers[:3]
# numbers[3:]
numbers[3:6]
````

### Common List Methods
### Common List Methods

| Method      | Purpose                                          | Example Usage              |
| ----------- | ------------------------------------------------ | -------------------------- |
| `append()`  | Adds one element to the end                      | `numbers.append(80)`       |
| `insert()`  | Adds an element at a specific position           | `numbers.insert(1, 65)`    |
| `extend()`  | Adds multiple elements                           | `numbers.extend([80, 90])` |
| `remove()`  | Removes the first matching value                 | `numbers.remove(50)`       |
| `pop()`     | Removes and returns an element (last by default) | `numbers.pop()`            |
| `clear()`   | Removes all elements                             | `numbers.clear()`          |
| `sort()`    | Sorts the list in ascending order by default     | `numbers.sort()`           |
| `reverse()` | Reverses the order of elements                   | `numbers.reverse()`        |
| `count()`   | Counts occurrences of a value                    | `numbers.count(50)`        |
| `index()`   | Returns the position of the first matching value | `numbers.index(50)`        |

Note: Most methods that modify a list (`append()`, `insert()`, `extend()`, `remove()`, `sort()`, `reverse()`, and `clear()`) modify the original list and return `None`. `pop()` returns the removed element, while `count()` and `index()` return values without modifying the list.
