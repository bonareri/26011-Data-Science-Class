## Introduction to Control Flow 

Python normally executes instructions sequentially, from top to bottom

`Control flow` refers to the order in which instructions are executed in a program 

Python provides two important mechanisms for controlling program execution

- `Conditional statements`: Allow the program to make decisions
- `Loops`: Allow programs to repeat instructions

````python
student_name = "Brian"
score = 85

print(student_name)
print(score)
````

## Conditional Statements
`Conditionsl statements` allow a program to execute different instructions depending on whether a condition evaluates to `True` or `False`

### The `if` Statement

An `if` statement executes a block of code only when the condition is `True`

**Syntax**
```python
if condition:
    # Instructions to execute when the condition is True
```

````python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote.")
````

**Key concepts**
- A `condition` - is an expression that evaluates to `True` or `False`
- The colon `(:)` - marks the beginning of the conditional block
- `Indentation` - identifies the instructions beloging to the condition
- If the condition is `False`, python skips the instructions inside the block

## The `if-else` Statement

Somethimes you might need a program to choose between two possible outcomes

An `if-else` statement provides two possible execution paths:
- The `if` block executes when the condition is `True`
- The `else` block executes when the condition is `False`

````python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
````

## Chained Conditionals `if-elif-else`

Allows you to choose from than two possibilities

90 - 100 - A 
80 - 89 - B
70 -79 - C
60 - 69 - D
BELOW 60 - F

````python
score = 85

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
elif score >= 60:
    print("Grade D")
else: 
    print("Grade F")
````

````python
score = 95 

if score >= 60:
    print("Grade D")
elif score >= 70:
    print("Grade C")
elif score >= 80:
    print("Grade B")
elif score >= 90:
    print("Grade A")
else: 
    print("Grade F")
````

Quick Note:

-  Python evaluates the conditions from top to bottom
- The first condition that evaluates to `True` executes, the remaining branches are skipped

## Nested Conditionals

A `Nested Conditional` is an `if` statement placed inside another conditional statement

Imagine this: A student wants to access an examination

The student must:
1. Be registered
2. Have paid the required fees

Both requirements must be satisfied

````python
registered = True
fees_paid = False
fees_pending = False

# Include the condition to check that the fee payment deadline has not passed

if registered:
    if fees_paid:
        print("You are registered and your fees are paid.")
    elif fees_pending:
        print("You are registered and your fees are pending.")
    else:
        print("You are registered but your fees are not paid.")
else:
    print("You are not registered.")
````

## Short-Circuit Evaluation and the Guardian Pattern 

Python evaluates logical expression from left to right

With `and`, if the first condition is `False`, python does not evaluate the remaining conditions

This is called `short-circuit evaluation`

It can help prevent errors when an operation depends on another condition being satisfied first

#### Prevent Division by Zero

````python
y = 0
x = 10

if y != 0 and x / y:
    print("Safe")
else:
    print("Division by zero is not allowed.")
````
