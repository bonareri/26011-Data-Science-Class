## Introduction to Control Flow 

Python normally executes instructions sequentially, from top to bottom

`Control flow` refers to the order in which instructions are executed in a program 

Python provides two important mechanisms for controlling program execution

- `Conditional statements`: Allow the program to make decisions
- `Loops`: Allow programs to repeat instructions

````python
student_name = "Brian"
score = 85

print(student_name)
print(score)
````

## Conditional Statements
`Conditionsl statements` allow a program to execute different instructions depending on whether a condition evaluates to `True` or `False`

### The `if` Statement

An `if` statement executes a block of code only when the condition is `True`

**Syntax**
```python
if condition:
    # Instructions to execute when the condition is True
```

````python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote.")
````

**Key concepts**
- A `condition` - is an expression that evaluates to `True` or `False`
- The colon `(:)` - marks the beginning of the conditional block
- `Indentation` - identifies the instructions beloging to the condition
- If the condition is `False`, python skips the instructions inside the block

## The `if-else` Statement

Somethimes you might need a program to choose between two possible outcomes

An `if-else` statement provides two possible execution paths:
- The `if` block executes when the condition is `True`
- The `else` block executes when the condition is `False`

````python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
````

## Nested Conditionals

A `Nested Conditional` is an `if` statement placed inside another conditional statement

Imagine this: A student wants to access an examination

The student must:
1. Be registered
2. Have paid the required fees

Both requirements must be satisfied

````python
registered = True
fees_paid = False
fees_pending = False

# Include the condition to check that the fee payment deadline has not passed

if registered:
    if fees_paid:
        print("You are registered and your fees are paid.")
    elif fees_pending:
        print("You are registered and your fees are pending.")
    else:
        print("You are registered but your fees are not paid.")
else:
    print("You are not registered.")
````

## Short-Circuit Evaluation and the Guardian Pattern 

Python evaluates logical expression from left to right

With `and`, if the first condition is `False`, python does not evaluate the remaining conditions

This is called `short-circuit evaluation`

It can help prevent errors when an operation depends on another condition being satisfied first

#### Prevent Division by Zero

````python
y = 0
x = 10

if y != 0 and x / y:
    print("Safe")
else:
    print("Division by zero is not allowed.")
````

## While Loop

A `while` loop executes a block of code as long as its condition remains `True`

Used when you want to repeat something as long a particular condition remains `True`

**Syntax**
```python
while condition:
    # Instructions to repeat
```

````python
# Print to numbers from 1 to 5 
count = 1
while count <= 5:
    print(count) 
    count += 1  
````

````python
# Print all even numbers between 2 and 10 

number = 2 

while number <= 10:
    print(number)
    number += 2
````

An `inifinite loop` occurs when a loop continues executing withouit reaching a termination condition

## For Loop

A `for` loop iterates over items in a sequence, such as list, string or range of numbers

Used when you know what you want to iterate through or how many times you want to repeat something.

**Syntax**
```python
for variable in sequence:
    # Instructions
```

````python
# Print the names of students
students = ["Wayne", "Mark", "Sahal", "Jeniffer"]

for pupil in students:
    print(pupil)
````

## Iterate through Strings

````python
name = "Wayne"

for letter in name:
    print(letter)
````

````python
# Count how many characters the word contains

name = "Wayne"
count = 0

for letter in name:
    print(letter)

print(count)
````

## The `break` Statement

The `break` statement immediately terminates a loop 

````python
numbers = [3, 7, 10, 15, 20]

for number in numbers:
    print(number)  

    if number == 10:
        print("Number found!")
        break
````

## The `Continue` Statement

The `continue` statement skips the remaining instruction in the current iteration and moves to the next iteration

````python
numbers = [3, 7, 10, 15, 20]

for number in numbers:
    if number == 10:
        continue

    print(number)
        
````

## Using `range()`

The `range()` function generates a sequence of integers

**Syntax**
```pyrhon
range(start, stop, step)
```

Parameters
- `start` - Starting value, Defaults to 0
- `stop` - Ending boundary, which is excluded
- `step` - Difference between consuctive values, Defaults to 1

````python
# Print numbers from 1 to 5

for number in range(1, 10, 2):
    print(number)
````

````python
scores = [85, 92, 78, 96, 88]

for score in scores:
    if score >= 90:
        print(f"High score: {score}")
````
