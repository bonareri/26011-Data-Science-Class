
# Values, Variables and Data Types

## Understanding Values

A **value** is a piece of information that a program can work with.

Every value in Python has a data type. A data type determines the kind of information represented and the operations that can be performed on it.

### Basic Python Data Types

| Data type | Description | Example |
|---|---|---|
| Integer (`int`) | Whole numbers | `25` |
| Float (`float`) | Decimal numbers | `89.5` |
| String (`str`) | Text | `"Brian"` |
| Boolean (`bool`) | True or False values | `True` |

**Key concept:** Quotation marks indicate that a value is a string, even when the characters inside them look like numbers.

## Understanding Variables

A **variable** is a named reference to a value stored in a program.

Variables allow us to store information and reuse it without repeatedly writing the original values.

```python
name = "Melody"
age = 24
score = 45.5
has_passed = True
```

```python
print(name, age, score, has_passed)
```

## Updating Variables

A variable can be assigned a new value

```python
age = 30
print(age)
```

## Inspect Data Types

Python provides the `type()` function to identify the data type of a value

```python
print(type(name))
print(type(age))
print(type(score))
print(type(has_passed))
```

## Variable Naming Rules

1. Names must with a letter or an underscore
2. Names cannot start with a number
3. Spaces are not allowed
4. Names can contain letters, numbers and underscores
5. Variable names are case-sensitive
6. Python reserved keywords cannot be used as variable names

Python uses `snake_case` for variable names

```python
and     continue  finally  is        raise
as      def       for      lambda    return
assert  del       from     None      True
async   elif      global   nonlocal  try
await   else      if       not       while
break   except    import   or        with
class   False     in       pass      yield
# The above keywords are reserved in Python and cannot be used as variable names.
```

```python
student_name  = "Asma"
second_score = 85.0
student_age = 24
class_name= "Data science"
```

## Type Casting
`Type casting` of converting a value from one data type to another

```python
age = "25"

age = int(age)

print(age + 5)
```

### Common Type Conversion Functions

- `int()` - convert to an integer
- `float()` - convert to a float
- `str()` - convert to a string
- `bool()` - convert to a boolean

## Python Operators

`Operators` allow us to perform calculations, compare values and make decisions

### Arithmetic Operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 5` |
| `-` | Subtraction | `10 - 5` |
| `*` | Multiplication | `10 * 5` |
| `/` | Division | `10 / 5` |
| `//` | Floor division | `10 // 3` |
| `%` | Modulus | `10 % 3` |
| `**` | Exponentiation | `2 ** 3` |


```python
assignment_score = 75
practical_score = 85
final_score = 65

total_score = assignment_score + practical_score + final_score
print("Total Score:", total_score)
```

```python
x = 10 
y = 3

print("Addition:", x + y)
print("Floor Division:", x // y)
print("Modulus:", x % y)
print("Exponentiation:", x ** y)
```

## Comparison Operators

Comparison operators compare two values and return a Boolean result (`True` or `False`)

- `>` Greater than
- `<` - Less than
- `>=` - Greater than or equal to
- `<=` - Less than or equal to
- `==` Equal to 
- `!=` Not equal to 

```python
average = 73

print(average >= 70)
```

## Logical Operators

Logical operators combine multiple conditions

- `and`- Both conditions must ne True
- `or` - Atleast one condition must be True
- `not` - Reverses the Boolean result

```python
assignment_score = 75
final_exam_score = 32

assignment_passed = assignment_score >= 50 
exam_passed = final_exam_score >= 50

overall_pass = assignment_passed and not exam_passed
print("Overall Pass:", overall_pass)
```

## Assignment Operators
`Assignment operators` update the value of an existing variable

- `+=` - x = x + value
- `-=` - x = x - value
- `*=` - x = x * value
- `/=`
- `//=`
- `%=`
- `**=`

```python
initial_total = 100

initial_total += 50
print("After addition:", initial_total)
```

```python
x = 10 
#x = x + 5
x += 5
print(x)
```

## Understanding Expressions

An `expression` is a combination of values, variables, operators or function calls that python evaluates to produce a result

```python
x = 2
y = 3
x + y
```

```python
10 + 5 * 2
```

```python
name = "Jeniffer" + "Melody"
print(name)
```

### Operator Precedence
Operator precedence determines the order in which python evaluates operations in an expression

Parenthesis are useful for controlling the evaluation order 

```python
(10 + 5) * 2
```

```python
(2 + 3) * 4 
```
