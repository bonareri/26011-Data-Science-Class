## Functions

A `function` is a reusable block of code designed to perform a specific task

Instead of writing the same instructions repeatedly, we define them once and call the function whenever we need them.

Functions help us organise programs into smaller, manageable and reusable pieces.

**Function Syntax**

Python uses the `def` keyword to define a function

```python
def function_name():
    # Instructions to execute

### Functions Without Parameters

A function without parameters does not receive values from the code calling 

````python
def greet():
    print("Welcome to Python Class!")

greet()
````

### Functions With Parameters

Sometimes a function needs information to perform a task

We can provide that information through `parameters`

````python
def greet_student(name):
    print("Welcome", name)

greet_student("Mark")
greet_student("Wayne")
````

Here `name` is a parameter. It allows the function to work different values.

````python
# Calculate the Area of a rectangle

def calculate_area(length, width):
    area = length * width
    print("Area", area)

calculate_area(10, 5)
calculate_area(5, 4)
````

## Parameters vs Arguments

- `Parameters` - are variable names specified when defining the function
- `Arguments` - are actual values supplied when calling the function

````python
def calculate_area_square(side_length):
    area = side_length ** 2
    print("Area of the square: ", area)

calculate_area_square(5)
````

- `side_length` - parameter
- `5` - the argument

## Types of Arguments 

### Positional Arguments

Values assigned according to their position

````python
def student_details(name, age):
    print("Student Name: ", name)
    print("Student Age: ", age)

student_details("Nasra", 25)
````

### Keyword Arguments

Values are assigned using the parameter names

````python
def student_details(name, age):
    print("Student Name: ", name)
    print("Student Age: ", age)

student_details(age=22, name="Nasra")
````

### Default Arguments

A parameter can have a default value that is used when no arguemnt is supplied

````python
def greet_zindua_student(name, age, course="Data Science"):
    print(f"{name}, is studying, {course}, and you are {age}, years old"  )

greet_zindua_student("Wayne", 25,  "Data Analysis",)
````

### Variable-Length Arguments

Sometimes we do not know how many arguments a function will receive

Python provides *args and **kwargs

`*args`

Allows a function to receive multiple positional arguments

````python
def calculate_total(*numbers):
    total = sum(numbers)
    print(f"The total is: {total}")


calculate_total(10, 20, 30, 40, 50)
````

````python
def calculate_average(*marks):
    average = sum(marks) / len(marks)
    average_rounded = round(average, 2)
    print(f"The average is: {average_rounded}")

calculate_average(23, 27, 45)
````

### Returning Values:
#### `print()` vs `return`

A function can perform a task and display the information, or it can calculate something and send the result back to the program that called it

### Using `print()`

`print()` displays information on the screen but does not return the calculated value for further use

````python
def calculate_sum(*numbers):
    total = sum(numbers)
    print(total)

calculate_sum(200, 300)
````

### Using `return`

The `return` statement sends a value back to the calling code.

````python
def calculate_sum(*numbers):
    total = sum(numbers)
    return total

result = calculate_sum(20, 20)
print(result)
````

We can now store the result and use it in another calculation

````python
final_amount = result + 100

print(final_amount)
````

### Fruitful and Void Functions

#### Fruitful Functions
- Return a value usng `return`.
- Results can be stored in variables
- Can be used in other calculations 

````python
def calculate_discount(price, discount):
    discount_amount = price * (discount/100)
    final_price = price - discount_amount

    return final_price

result = calculate_discount(5000, 15)
print("Final price: ", result)
````

### Void Functions

- Do not excplicilty return a value
- Usually perform an action
- Often display information or perform a task

````python
def display_message():
    print("Welcome to Python Class")

display_message()
````

## Making Random Results Reproducible

Randomness can make experiments difficult to reproduce

We can use `random.seed()` to make random results reproducible

````python
random.seed(5)
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(numbers)
````

### Where Are Random Functions Useful in Data Science

- Randomly selecting samples from a population.
- Shuffling datasets before model training.
- Splitting data into training and testing sets.
- Simulating experiments.
- Generating sample data for testing.

## Lambda Functions

A `lambda function` is a small anonymus function defined using the `lambda` keyword

Unlike regular functions, lambda functions are written in a single expression and do not require a function name in their definition

**Syntax**
```python
lambda arguments: expression
```

#### Example: Squaring a Number

````python
# Using a regular function
def square(number):
    return number ** 2

result = square(4)
print("Result:", result)
````

````python
# Using a lambda function

square = lambda number: number ** 2

print(square(4))
````

````python
# Using lambda with multiple arguments

add = lambda a, b: a + b 

add(10, 20)
````

### Lambda With `map()`

The `map()` function applies a function to every item in an iterable

````python
numbers = [1, 2, 3, 4, 5]

square = list(map(lambda digit: digit ** 2, numbers))

print(square)
````

### Lambda With `filter()`

````python
numbers = [10, 15, 20, 25, 30, 35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
````

### Why Should We Use Lambda Functions?

- For short operations
- When passing a simple function to another function
- When sorting data using a particular column
- When using functions such as `map()` and `filter()`
