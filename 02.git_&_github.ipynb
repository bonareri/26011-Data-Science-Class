## 3. What Is Version Control?

Version control is a system that helps you track changes made to files over time.

It allows you to:

- Keep a history of your changes.
- Return to an earlier version of your work.
- Identify what changed and when.
- Work on different versions of a project.
- Collaborate without constantly sharing files manually.

**Key idea:** Version control helps you manage the history of your work.

## 4. What Is Git?

Git is a distributed version control system that tracks changes in your files.

With Git, you can:

- Track changes to your code.
- Save important versions of your project.
- Compare different versions.
- Create separate branches to experiment safely.
- Combine changes from different contributors.

Git works locally on your computer, even without an internet connection.

## 5. What Is GitHub?

GitHub is an online platform that hosts Git repositories.

It allows you to:

- Store your projects online.
- Share your work with others.
- Collaborate with other developers.
- Review and discuss code.
- Showcase your projects in a portfolio.

**Think of it this way:**

Git is the tool that tracks changes, while GitHub is the platform where you can host, share and collaborate on those projects.

## 6. Git vs GitHub

| Git | GitHub |
|---|---|
| Version control system | Online hosting platform |
| Runs locally on your computer | Accessible through the internet |
| Tracks changes | Hosts and shares repositories |
| Supports commits and branches | Supports pull requests and collaboration |
| Can work offline | Requires internet for online features |

**Quick check:** Can you use Git without GitHub?

Yes. Git works independently on your computer.

## 7. Important Git Terminology

Before using Git, understand these concepts:

- **Repository:** A project folder whose history is tracked by Git.
- **Working directory:** The files you are currently working on.
- **Staging area:** An area where you select changes to include in your next commit.
- **Commit:** A snapshot of selected changes.
- **Branch:** A separate line of development.
- **Remote repository:** A repository hosted somewhere other than your local computer, such as GitHub.
- **Clone:** Creating a local copy of an existing remote repository.
- **Push:** Uploading local commits to a remote repository.
- **Pull:** Fetching and integrating changes from a remote repository.

## 8. Understanding the Git Workflow

Git follows a process when you save changes.

1. **Working directory:** Make changes to your files.
2. **Staging area:** Select the changes you want to save.
3. **Local repository:** Commit the selected changes.
4. **Remote repository:** Push your commits to GitHub.

The workflow:

`Working Directory → Staging Area → Local Repository → GitHub`

**Remember:** Saving a file is not the same as committing it, and committing a file is not the same as pushing it to GitHub.

## 12. Install Git on Your Computer

Now that you have created a repository online, you need Git on your computer to manage your project locally.

Download Git:

```bash
git --version
```

If Git is installed correctly, you should see its version number.

Git needs to know who is making commits.

Run:

```bash
git config --global user.name "Your Name"
```

## 13. Clone Your GitHub Repository

# PART THREE: WORKING WITH GIT LOCALLY

## 14. Understanding Your Local Repository

Inside your project folder, Git maintains a hidden directory called `.git`.

This directory stores information about the repository, including its history, branches and configuration.

To check the current state of your repository, run:

```bash
git status
```

You should see something similar to:

```text
On branch main
Your branch is up to date with 'origin/main'.

nothing to commit, working tree clean
```

What does this mean?

- You are currently working on the `main` branch.
- Your local repository is synchronized with GitHub.
- There are no uncommitted changes.

## 15. Create Your First Project Files

We will now add files to our student performance analysis project.

Inside your project directory, create a Python file:

```bash
touch analysis.py
```

Open the file in your code editor and add:

```python
print("Student Performance Analysis")

students = 50
average_score = 72.5

print(f"Number of students: {students}")
print(f"Average score: {average_score}")
```

Save the file.

You can also create another file:

```bash
touch requirements.txt
```

Your project should now look like this:

```text
student-performance-analysis/
│
├── README.md
├── analysis.py
└── requirements.txt
```

**Think:** Have these new files automatically appeared on GitHub?

No. They currently exist in your local working directory.

We need to use Git to track and save the changes.

## 16. Tracking Changes with Git

Git allows you to track changes made to your project.

First, check the repository status:

```bash
git status
```

You should see that `analysis.py` and `requirements.txt` are untracked files.

**What does untracked mean?**

It means Git can see the files, but they have not yet been added to the staging area.

### Stage a specific file

```bash
git add analysis.py
```

### Stage all changes

```bash
git add .
```

Check the status again:

```bash
git status
```

The files should now appear under changes to be committed.

**Think:** What is the difference between creating a file and staging a file?

## 17. Saving Changes with Commits

A commit records a snapshot of your staged changes.

To create a commit:

```bash
git commit -m "Add initial analysis script"
```

A good commit message should explain what changed.

Examples:

- `Add data cleaning script`
- `Update project documentation`
- `Fix missing values in dataset`
- `Add exploratory data analysis`

Avoid vague messages such as:

- `Update`
- `Changes`
- `Final work`

**Activity:** Add another line to your Python file, stage the changes and create another commit.

## 18. Viewing Your Project History

Git allows you to review the history of your project.

View all commits:

```bash
git log
```

View a simplified history:

```bash
git log --oneline
```

Compare changes that have not been staged:

```bash
git diff
```

Compare staged changes:

```bash
git diff --staged
```

**Think:** Why might a data analyst need to review previous versions of a Python script or notebook?

## 19. Pushing Changes to GitHub

You have created files, staged them and committed your changes.

However, your changes are still stored locally.

How do you send them to GitHub?

You use `git push`.

Run:

```bash
git push
```

Open your GitHub repository and refresh the page.

You should now see your files.

**Important:** If GitHub requests authentication, follow the browser sign-in or credential manager prompts. Your normal GitHub account password is not used directly as an HTTPS Git password.

### Understanding the difference

- `git add` selects changes.
- `git commit` saves a snapshot locally.
- `git push` uploads committed changes to GitHub.

**Quick activity:** Change the average score in your Python file, commit the change and push it to GitHub.

# PART FOUR: BRANCHES AND COLLABORATION

## 20. Understanding Branches

A branch allows you to work on a separate version of your project without directly changing the main branch.

For example:

- `main`: The stable version of your project.
- `data-cleaning`: Work on cleaning the dataset.
- `exploratory-analysis`: Work on exploratory data analysis.
- `visualizations`: Work on creating charts.

### Create a new branch

```bash
git switch -c data-cleaning
```

This command creates a branch and switches to it.

List available branches:

```bash
git branch
```

Switch back to the main branch:

```bash
git switch main
```

**Activity:** Create a branch called `exploratory-analysis` and switch to it.

**Think:** Why would a data scientist want to experiment with a new analysis without immediately changing the stable version of a project?

## 21. Working on a Branch

Let's assume you are working on the `data-cleaning` branch.

Switch to it:

```bash
git switch data-cleaning
```

Open `analysis.py` and add:

```python
print("Data cleaning will happen here.")
```

Save the file.

Check the changes:

```bash
git status
```

Stage and commit:

```bash
git add .
git commit -m "Add data cleaning section"
```

Push the branch to GitHub:

```bash
git push -u origin data-cleaning
```

Visit GitHub and check whether the new branch is visible.

**What happened?**

Your new work has been pushed to GitHub without directly changing the `main` branch.

## 22. Merging Branches

After completing work on a branch, you may want to combine those changes with the main branch.

This process is called merging.

First, switch to the main branch:

```bash
git switch main
```

Then merge your branch:

```bash
git merge data-cleaning
```

Push the updated main branch:

```bash
git push
```

Your changes are now included in the main branch.

**Think:** Why is it useful to develop a new feature on a separate branch before merging it into the main project?

## 23. Understanding Pull Requests

A pull request is a GitHub feature that allows you to propose changes and request that they be reviewed before being merged into another branch.

For example:

1. You create a branch called `data-cleaning`.
2. You make changes and push the branch to GitHub.
3. GitHub allows you to open a pull request.
4. Another developer reviews your changes.
5. The changes are approved and merged into `main`.

Pull requests are particularly useful when working in teams.

**Important distinction:** A Git merge combines changes locally, while a GitHub pull request provides a platform for reviewing and discussing proposed changes before merging.

## 25. What Is the Difference Between Git Init and Git Clone?

These commands are often confused, but they serve different purposes.

**git init**

Creates a new Git repository inside an existing local project folder.

Example:

```bash
mkdir my-project
cd my-project
git init
```

Use this when you already have a local project and want to start tracking it with Git.

**git clone**

Creates a local copy of an existing remote repository.

Example:

```bash
git clone https://github.com/USERNAME/my-project.git
```

Use this when the repository already exists on GitHub and you want to work on it locally.

**Remember:**

- `git init` starts a new local repository.
- `git clone` copies an existing repository.

Do not run `git init` inside a repository you have already cloned.

# PART SIX: GIT AND GITHUB IN DATA SCIENCE

## 26. Why Git and GitHub Matter in Data Science

As a data professional, you will work with:

- Python scripts and notebooks.
- Data cleaning and preprocessing pipelines.
- Machine learning models.
- Project documentation.
- Team-based projects.

Git and GitHub help you keep your work organized, reproducible and accessible.

### Example: A Data Science Repository Structure

A typical data science repository might look like this:

```text
student-performance-analysis/
│
├── data/
│   └── student_performance.csv
│
├── notebooks/
│   └── exploratory_analysis.ipynb
│
├── src/
│   └── data_cleaning.py
│
├── README.md
├── requirements.txt
└── .gitignore
```

### What is a README file?

A README file introduces your project.

It typically contains:

- Project title.
- Project description.
- Problem statement.
- Dataset information.
- Tools and technologies used.
- Instructions for running the project.
- Key findings.

### What is a .gitignore file?

A `.gitignore` file tells Git which files and directories should not be tracked.

For example:

```text
__pycache__/
.ipynb_checkpoints/
.env
.venv/
*.pyc
```

This helps prevent unnecessary files and sensitive configuration files from being uploaded.

**Important:** Do not upload passwords, API keys, confidential information or private datasets to public repositories.
