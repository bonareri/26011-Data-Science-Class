### Build a Command-Line Quiz Application

The application should:
 
- display multiple-choice questions
- mix question order.
- accept answers from users
- checks whether the answers are correct
- keep track of scores
- provide feedback based on the users performace

### Import the required Module

`random` module to shuffle the order of the questions

```python
import random
```

### Create the Questions

Store the quiz questions in a list of dictionaries

Every dictionary should contain:
- The question
- Four answer choices: A, B, C and D
- The correct answer

```python
questions = [
    {
        "question":"Which keyword is used to define a function in python?",
        "A": "function",
        "B": "def",
        "C": "fun",
        "D": "define",
        "answer": "B"
    },
    {
        "question":"Which data type is used to store multiple items in an ordered collection?",
        "A": "Integer",
        "B": "String",
        "C": "List",
        "D": "Boolean",
        "answer": "C"
    },
    {
        "question":"Which loop is commonly used to iterate over a sequence?",
        "A": "if",
        "B": "def",
        "C": "try",
        "D": "for",
        "answer": "D"
    },
]
questions
```

### Shuffle the Questions

Mix the order of the questions 

```python
random.shuffle(questions)
questions
```

### Initialise the Score

The score should increase every time the user selects the correct answer

```python
score = 0
```

### Display the questions to the user

Use `for` loop to go through the questions and display them one at a time

For each question, display:
- The question
- The answer choices

```python
for question in questions:
    print("\n" + question["question"])

    print("A", question["A"])
    print("B", question["B"])
    print("C", question["C"])
    print("D", question["D"])
```

### Accept and Validate User Input

Use `while` loop and `try/except` to handle invalid input

```python
while True:
    try:
        answer = input("Enter your answer (A/B/C/D):").upper().strip()

        if answer not in ["A", "B", "C", "D"]:
            raise ValueError("Please enter A, B, C or D.")
        break 
    except ValueError as error:
        print(error)
```

### Check the Answer and Update the Score

Compare the user's answer with the correct answer stored in the dictionary

If the answer is correct:
- Display a message confirming the correct answer
- Increase the score by 1

If the answer is incorrect:
- Display an incorrect-answer message
- Show the correct answer

```python
if answer == question["answer"]:
    print("Correct")
    score += 1
else:
    print("Incorrect")
    print("Correct answer:", question["answer"])
```

### Calculate the Final Score

```python
total_questions = len(questions)
percentage = (score / total_questions) * 100

print("\nQuiz Results")
print("Your score:", score, "/", total_questions)
print("Percentage:", round(percentage, 2), "%")
```

### Provide Performance Feedback

- 80 % and above - Excellent
- 50% - 79% - Good
- Below 50 % - Try Again

```python
if percentage >= 80:
    print("Excellent")
elif percentage >= 50:
    print("Good")
else:
    print("Try Again!")
```
